```python
from game.game_engine import GameEngine
from agents.minimax_agent import MinimaxAgent
from agents.alphabeta_agent import AlphaBetaAgent

# Create identical initial game states
game = GameEngine(board_size=4, win_length=4)

# Initialize both agents with the same depth
minimax = MinimaxAgent(depth=4)
alphabeta = AlphaBetaAgent(depth=4)

# Run Minimax
minimax_action = minimax.choose_action(game)

# Run Alpha-Beta
alphabeta_action = alphabeta.choose_action(game)

# Display results
print("========== COMPARISON ==========")

print("\nMinimax:")
print("Action:", minimax_action)
print("Nodes:", minimax.nodes_visited)
print("Time:", minimax.decision_time)

print("\nAlpha-Beta:")
print("Action:", alphabeta_action)
print("Nodes:", alphabeta.nodes_visited)
print("Time:", alphabeta.decision_time)
print("Pruned branches:", alphabeta.pruned_branches)
print("Cutoffs:", alphabeta.cutoffs)

print("\n========== ANALYSIS ==========")

if minimax.nodes_visited > 0:
    reduction = (
        (minimax.nodes_visited - alphabeta.nodes_visited)
        / minimax.nodes_visited
    ) * 100

    print(f"Node reduction: {reduction:.2f}%")

print("Same action:", minimax_action == alphabeta_action)
```

```python
from game.game_engine import GameEngine
from agents.mcts_agent import MCTSAgent

# Initialize game
game = GameEngine(board_size=4, win_length=4)

# Initialize MCTS
mcts = MCTSAgent(
    simulations=1000,
    exploration=1.4142135623730951,
    seed=42
)

# Find best action
action = mcts.choose_action(game)

print("Best action:", action)

print("\nMCTS Statistics:")

for key, value in mcts.last_search_stats.items():
    print(f"{key}: {value}")
```

```python
print("Selected action:", action)
print("Is legal:", action in game.legal_actions())

print("\nBoard after MCTS search:")
game.render()
```
