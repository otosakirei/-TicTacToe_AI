```python
import pandas as pd
import numpy as np

matches = pd.read_csv(
    r"D:\TTNTNC\TicTacToe_AI\experiments\results\match_results.csv"
)

agents = [
    "Random",
    "Heuristic",
    "Minimax",
    "Alpha-Beta",
    "MCTS",
    "DQN"
]

# Ma trận Score của AI hàng khi gặp AI cột
head_to_head = pd.DataFrame(
    np.nan,
    index=agents,
    columns=agents
)

for agent_a in agents:
    for agent_b in agents:

        if agent_a == agent_b:
            continue

        games = matches[
            (
                (matches["agent_x"] == agent_a) &
                (matches["agent_o"] == agent_b)
            ) |
            (
                (matches["agent_x"] == agent_b) &
                (matches["agent_o"] == agent_a)
            )
        ]

        wins = 0
        draws = 0

        for _, game in games.iterrows():
            if game["winner"] == agent_a:
                wins += 1
            elif game["winner"] == "DRAW":
                draws += 1

        head_to_head.loc[agent_a, agent_b] = (
            (wins + 0.5 * draws) / len(games)
            if len(games) else np.nan
        )

print("MA TRẬN ĐỐI ĐẦU – SCORE")
display(head_to_head.round(3))
```

```python
first_move_results = []

for agent in agents:

    x_games = matches[matches["agent_x"] == agent]
    o_games = matches[matches["agent_o"] == agent]

    x_wins = (x_games["winner"] == agent).sum()
    x_draws = (x_games["winner"] == "DRAW").sum()

    o_wins = (o_games["winner"] == agent).sum()
    o_draws = (o_games["winner"] == "DRAW").sum()

    first_move_results.append({
        "Agent": agent,
        "X_Games": len(x_games),
        "X_Wins": x_wins,
        "X_Draws": x_draws,
        "X_Score": (x_wins + 0.5*x_draws) / len(x_games),
        "O_Games": len(o_games),
        "O_Wins": o_wins,
        "O_Draws": o_draws,
        "O_Score": (o_wins + 0.5*o_draws) / len(o_games)
    })

first_move_df = pd.DataFrame(first_move_results)

display(first_move_df.round(3))
```

```python
moves = pd.read_csv(
    r"D:\TTNTNC\TicTacToe_AI\experiments\results\move_results.csv"
)

timing_summary = moves.groupby("agent")["decision_time"].agg(
    ["count", "mean", "median", "max"]
)

timing_summary["p95"] = (
    moves.groupby("agent")["decision_time"]
    .quantile(0.95)
)

timing_summary = timing_summary.rename(columns={
    "count": "Total Moves",
    "mean": "Average Time",
    "median": "Median Time",
    "max": "Max Time",
    "p95": "P95 Latency"
})

display(timing_summary.round(6))
```

```python
import pandas as pd

matches = pd.read_csv(
    r"D:\TTNTNC\TicTacToe_AI\experiments\results\match_results.csv"
)

agents = [
    "Random",
    "Heuristic",
    "Minimax",
    "Alpha-Beta",
    "MCTS",
    "DQN"
]

pairwise_results = []

for i, agent_a in enumerate(agents):
    for agent_b in agents[i + 1:]:

        games = matches[
            (
                (matches["agent_x"] == agent_a) &
                (matches["agent_o"] == agent_b)
            ) |
            (
                (matches["agent_x"] == agent_b) &
                (matches["agent_o"] == agent_a)
            )
        ]

        wins_a = (games["winner"] == agent_a).sum()
        wins_b = (games["winner"] == agent_b).sum()
        draws = (games["winner"] == "DRAW").sum()

        score_a = (wins_a + 0.5 * draws) / len(games)

        pairwise_results.append({
            "Agent A": agent_a,
            "Agent B": agent_b,
            "Games": len(games),
            "A Wins": wins_a,
            "Draws": draws,
            "B Wins": wins_b,
            "A Score": score_a,
            "B Score": 1 - score_a
        })

pairwise_df = pd.DataFrame(pairwise_results)

display(pairwise_df.round(3))
```
